```python
import pandas as pd 
```

```python
def clean_regulation_text(df):
    """
    Clean scraped regulation text and separate regulation notes.

    For each row in the "text" column:
    1. Remove everything before and including the first occurrence of
       "State Regulations".
    2. Create a new "notes" column containing everything after "Notes".
    3. Remove "Notes" and everything after it from the "text" column.

    The original spacing, tabs, and line breaks are preserved.

    Parameters
    ----------
    df : pandas.DataFrame
        DataFrame containing a "text" column.

    Returns
    -------
    pandas.DataFrame
        A copy of the DataFrame with a cleaned "text" column and a new
        "notes" column.
    """

    df = df.copy()

    def clean_row(text):
        if pd.isna(text):
            return "", ""

        text = str(text)

        # Remove everything before and including "State Regulations"
        if "State Regulations" in text:
            text = text.split("State Regulations", 1)[1]

        # Split regulation text and notes
        if "Notes" in text:
            regulation_text, notes = text.split("Notes", 1)
        else:
            regulation_text = text
            notes = ""

        return regulation_text, notes

    cleaned = df["text"].apply(clean_row)

    df["text"] = cleaned.apply(lambda x: x[0])
    df["notes"] = cleaned.apply(lambda x: x[1])

    return df
```

## This is to do a singule state clean

```python
clean_up_state= "california" 

pre_df = pd.read_csv(f"states_raw_data/cornell_{clean_up_state}_regulations.csv")
```

```python
df = clean_regulation_text(pre_df)
```

```python
df.to_csv(f"states_clean_data/{clean_up_state}_cleaned_regulation_text.csv")
```

## This is to clean multiple states

```python
states_to_clean = ['california', 'new-york', 'new-mexico', 'arizona', 'texas'] 

for state in states_to_clean: 
    print(state)
    pre_df = pd.read_csv(f"states_raw_data/cornell_{state}_regulations.csv")
    df = clean_regulation_text(pre_df)
    df.to_csv(f"states_clean_data/{state}_cleaned_regulation_text.csv")
```
